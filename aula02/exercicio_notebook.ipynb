Importando o Seaborn para plotar os graficos

```python
import seaborn as sns
```

Gerei um grafico para as categorias e suas vendas 

```python
grafico = sns.histplot(data=df,x="categoria",kde=True,bins=20)
grafico.set_title("Produtos mais vendidos")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Quantidade")
```

```python
grafico = sns.boxplot(data=df,x="categoria",y="preco_unitario")
grafico.set_title("Preço por categoria")
```
